```python
#1번
import os, requests
from dotenv import load_dotenv

load_dotenv()
KEY: str = os.environ["KOREAN_DICT_KEY"]
def search_word(q: str, num: int = 10, start: int = 1) -> dict:
    params = {
        "key" : KEY, 
        "q" : q,
        "req_type" : "json",
        "num" : num,
        "start" : start,
        "type1": "word",
    }

    r = requests.get("https://opendict.korean.go.kr/api/search", params=params, timeout = 10)
    r.raise_for_status()
    data = r.json()
    return data
    
```

```python
import json
data = search_word("김치")
print(json.dumps(data, ensure_ascii=False, indent=2)[:400])
```

```python
items = data["channel"]["item"]
print(f"총{data['channel']['total']}건, 이 페이지{len(items)}건")

for it in items[:5]:
    word = it["word"]
    pos = it["sense"][0].get("pos", "품사 없음")
    sense = it["sense"][0] ["definition"]
    print(f"{word}({pos}): {sense[:40]}")
```

```python
#2
import time

words : list[str] = ["김치", "라면", "만두", "김밥", "국수", "떡볶이", "불고기", "비빔밥"]


for w in words:
    data = search_word(w)
    print(f"{w}:{data['channel']['total']}건")
    time.sleep(0.3)
```

```python
import collections

lst = []

for w in words:
    time.sleep(0.3)
    data = search_word(w)
    items = data["channel"]["item"]
    
    for item in items:
        pos = item["sense"][0].get("pos") or "(미상)"
        lst.append(pos)
counter = collections.Counter(lst)
for pos, count in counter.most_common(3):
    print(f"{pos}: {count}건")
```

코드설명

1. 리스트를 순회하면서 검색어의 전체 결과 수를 프린트한다.

2.빈 리스트에 pos를 리스트를 순회하면서 append한다. 이때 pos는 .get을 통해 품사를 반환받는다. 혹은 없을 경우 미상을 받는다. collections.Counter를 이용하여 리스트 형식으로 각 품사의 횟수를 세고 counter.mostcommon(3) 을 이용하여 가장 많이 등장한 품사 세개를 센다.

가장 흔한 품사는 명사이며 이는 위의 단어들의 가장 대표되는 의미가 명사들이기 때문이다.
