```python
import numpy as np
import pandas as pd

works = pd.DataFrame({
    '작가': ['김유정', '김유정', '김유정',
           '현진건', '현진건',
           '이상', '이상',
           '나도향', '나도향',
           '채만식', '이효석'],
    '작품': ['봄봄', '동백꽃', '만무방',
           '운수 좋은 날', 'B사감과 러브레터',
           '날개', '권태',
           '벙어리 삼룡이', '물레방아',
           '레디메이드 인생', '메밀꽃 필 무렵'],
    '연도': [1935, 1936, 1934,
           1924, 1925,
           1936, 1937,
           1925, 1925,
           1934, np.nan],   # 이효석 '메밀꽃 필 무렵' 발표연도 누락
    '글자수': [12500, 9800, 13600,
            11200, 7400,
            18300, 14100,
            16800, 14500,
            22000, 11800],
})

authors = pd.DataFrame({
    '작가': ['김유정', '현진건', '이상', '나도향', '채만식', '이효석', '염상섭'],
    '생년': [1908, 1900, 1910, 1902, 1902, 1907, 1897],
    '몰년': [1937, 1943, 1937, 1926, 1950, 1942, 1963],
})

works
```

```python
#1
#a)
print(works.dtypes)
print(works.describe())
```

글자수에는 결측치 값이 없어서 정수형이지만 연도에는 NaN이 섞여있어서 그 열 전체를 float로 처리한다.

```python
#b)
print(works.isna().sum())
```

연도 열에 결측치가 하나 있다.

```python
#c)
works2 = works.copy()
works2['연도'] = works2['연도'].fillna(1936).astype(int)
print(works2.dtypes)
print(works2.tail(3))
```

```python
#2

after_1930 = works2[works2['연도'] >= 1930]
print(after_1930)

kim_or_lee = works2[(works2['작가'].isin(['김유정', '이상']))]
print(kim_or_lee)
```

```python
def categorize(n: int) -> str:
    if n < 10000:
        return '짧음'
    elif n < 15000:
        return '보통'
    else:
        return '긴'
    
works2['분량'] = works2['글자수'].apply(categorize)
print(works2[['작품', '글자수', '분량']])
    
```

```python
works2.sort_values(by=['연도', '글자수'], ascending=[True, False])
```

```python
#3
works2.groupby('작가')['글자수'].agg(['mean', 'count','sum'])
```

```python
works2['분량'].value_counts()
```

```python
works2.groupby(['작가', '분량']).size()
```

김유정 보통 조합이 2개로 가장 많은 조합이다. 김유정은 길지도 짧지도 않은 정도의 작품을 많이 썼음을 시사한다.

```python
#4

full = pd.merge(works2, authors, on='작가', how='left')
print(full)
full.shape
```

포함되지 않는다. how=left는 works2에 있는 키만 보존하기 때문이다.

```python
full['집필 당시 나이'] = full['연도'] - full['생년']
full[['작가', '작품', '연도', '생년', '집필 당시 나이']]
```

```python
full.groupby('작가')['집필 당시 나이'].mean().sort_values()
```

가장 낮은 평균 집필 나이는 나도향 작가이다. 평균 23세에 작품을 발표했다. 가장 늦은 평균나이는 채만식이다.
